```python
import pandas as pd
import numpy as np
!pip install pgmpy
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.estimators import MaximumLikelihoodEstimator
from pgmpy.inference import VariableElimination

# Step 1: Define the structure of the Bayesian Network
model = DiscreteBayesianNetwork([
    ('Infection', 'Fever'),
    ('Infection', 'Cough'),
    ('Infection', 'Blood_Test'),
    ('Fever', 'Fatigue')
])

# Step 2: Create a Sample WHO-like dataset
data_dict = {
    'Infection': np.random.choice(['Viral', 'Bacterial', 'Fungal'], 100, p=[0.5, 0.3, 0.2]),
    'Fever': np.random.choice(['Yes', 'No'], 100, p=[0.7, 0.3]),
    'Cough': np.random.choice(['Yes', 'No'], 100, p=[0.6, 0.4]),
    'Blood_Test': np.random.choice(['HighWBC', 'NormalWBC'], 100, p=[0.4, 0.6]),
    'Fatigue': np.random.choice(['Yes', 'No'], 100, p=[0.5, 0.5])
}

# Convert to DataFrame
df = pd.DataFrame(data_dict)

# Step 3: Train the Bayesian Network
model.fit(df, estimator=MaximumLikelihoodEstimator)

# Step 4: Perform Inference
inference = VariableElimination(model)

# Step 5: Query the model for a diagnosis
evidence = {
    'Fever': 'Yes',
    'Cough': 'Yes',
    'Blood_Test': 'HighWBC'
}

result = inference.map_query(variables=['Infection'], evidence=evidence)

# Display the result
print("\nMost Likely Infection Diagnosis:")
print(f"Infection Type: {result['Infection']}")
```

```python
# Install pgmpy
!pip install pgmpy -q

# Import libraries
import pandas as pd
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.estimators import MaximumLikelihoodEstimator
from pgmpy.inference import VariableElimination

# -----------------------------
# Step 1: Create Dataset
# -----------------------------
data = {
    'StudyHours': ['High', 'High', 'Low', 'Low', 'High', 'Low'],
    'Attendance': ['High', 'Low', 'High', 'Low', 'High', 'Low'],
    'Pass': ['Yes', 'Yes', 'Yes', 'No', 'Yes', 'No']
}

df = pd.DataFrame(data)

print("Dataset:")
display(df)

# -----------------------------
# Step 2: Define Bayesian Network
# -----------------------------
model = DiscreteBayesianNetwork([
    ('StudyHours', 'Pass'),
    ('Attendance', 'Pass')
])

# -----------------------------
# Step 3: Train Model
# -----------------------------
model.fit(df, estimator=MaximumLikelihoodEstimator)

# -----------------------------
# Step 4: Inference
# -----------------------------
inference = VariableElimination(model)

# Example Query
evidence = {
    'StudyHours': 'Low',
    'Attendance': 'Low'
}

result = inference.query(variables=['Pass'], evidence=evidence)

print("\nProbability of Passing (Low Study, Low Attendance):")
print(result)

# MAP prediction
map_result = inference.map_query(variables=['Pass'], evidence=evidence)

print("\nMost Likely Outcome:")
print(map_result)
```
